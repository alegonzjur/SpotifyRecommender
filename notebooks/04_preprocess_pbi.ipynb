# 4 - Preprocesamiento de Campos para Power BI

```python
import pandas as pd
```

```python
df = pd.read_csv('../data/processed/tracks_clustered.csv')
# Mostramos las columnas.
df.columns
```

```python
# Mapeamos la columna Explicit para que sea entendible en PowerBI.
df['explicit_label'] = df['explicit'].map({True: 'Explícita', False: 'No explícita'})

# Duración en formato legible.
df['duration_min'] = (df['duration_ms'] / 60000).round(2)

# Asegurar categorías de popularidad.
df['popularity_category'] = pd.cut(df['popularity'],bins=[-1, 25, 50, 75, 100],labels=['Baja', 'Media', 'Alta', 'Viral'])

# Asegurar tipos correctos
df['popularity'] = df['popularity'].astype(int)
df['cluster'] = df['cluster'].astype(int)
```

```python
# Guardar versión final para Power BI
df.to_csv('../data/processed/tracks_powerbi.csv', index=False)
print(f"Dataset para Power BI guardado ✅")
print(f"Shape: {df.shape[0]:,} filas, {df.shape[1]} columnas")
print(f"\nColumnas finales:\n{df.columns.tolist()}")
```
